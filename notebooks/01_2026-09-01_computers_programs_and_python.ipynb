## Conceptual analogy


A restaurant kitchen provides a useful comparison. The appliances and counters are **hardware**. A recipe is like an **algorithm**: a plan that lists steps. A cook carrying out those steps is like the **CPU**. The counter is like short-term **memory** because ingredients being used now must stay close. The pantry is like **storage** because items remain there after the current task ends. Unlike a human cook, a computer does not fill in missing instructions through judgment or experience.


> This analogy provides an initial mental model. The definitions and Python examples below provide the precise technical meaning.


## Key terms

| Term | Plain-language meaning |
|---|---|
| **hardware** | The physical parts of a computer that can be touched. |
| **software** | Instructions and data that tell hardware what work to perform. |
| **CPU** | The component that executes program instructions. |
| **memory (RAM)** | Fast, temporary working space used by running programs. |
| **storage** | Long-term space for files and applications. |
| **algorithm** | A finite, ordered plan for solving a problem. |
| **program** | An algorithm expressed as instructions a computer can execute. |
| **Jupyter notebook** | An interactive `.ipynb` document that organizes text cells, code cells, and saved outputs in an ordered sequence. |
| **Google Colab** | A browser-based service for creating and running Jupyter notebooks without installing Python locally. |
| **Google Drive** | Persistent cloud storage used for personal notebook copies and course folders. |
| **runtime** | The temporary remote computer that executes a Colab notebook's code. |


## Explanation


### From an idea to a visible result

1. A user provides **input**, such as clicking a button or typing text.
2. A program's instructions and current data are loaded into memory.
3. The CPU executes instructions in order: fetch, interpret, and act.
4. The computer produces **output**, such as text, sound, or a saved file.
5. Saving moves information to storage so it remains after power is removed.

Python is a programming language we use to express instructions clearly. The Python interpreter translates and carries out those instructions. Python is not the CPU or the operating system; it is one layer in the larger system.

### What is a Jupyter notebook?

A **Jupyter notebook** is an interactive document, normally saved as an `.ipynb` file. It keeps related material in an ordered sequence of cells:

- **Text cells** contain headings, instructions, links, explanations, and reflections written with Markdown formatting.
- **Code cells** contain Python instructions that a connected runtime can execute.
- **Outputs** appear below code cells and may include printed text, tables, error messages, or other results. Saved outputs can be stored in the notebook file.

The notebook file is not the Python runtime. The file stores the cells and their order; the runtime performs the computation and temporarily remembers current variables. A notebook can therefore remain saved even after its runtime disconnects, but its code must be run again in a new runtime to recreate the working state.

### Where Google Colab fits

Google Colab displays a Jupyter notebook in a web browser and sends its Python code to a temporary remote runtime. The runtime executes the code and returns the output to the browser. The notebook file can be stored in Google Drive, but the runtime's variables and temporary files are separate. Saving the notebook preserves its cells; it does not preserve every temporary runtime file.

A dedicated Python installation, Jupyter installation, or virtual environment is therefore not required on a personal computer for this course. A current browser, internet access, and a working Google account with Drive access are required for the standard course workflow.

### Result and diagnosis

Python raises a `SyntaxError` because the opening parenthesis has no matching closing parenthesis. The idea is understandable to a person, but the written instruction is incomplete for Python.

**Error-reading procedure:** Read the final line of the message, locate the referenced line, and compare the value or structure Python received with the program's intended behavior.


### Correction steps

1. Read the final error line and locate the unmatched opening parenthesis.
2. Close the parenthesis so the instruction has a complete structure.
3. Run the corrected instruction and compare the output with the original intention.


```python
message = "A program is a sequence of instructions"
print(message)
```

### How the corrected example works

1. The text is data stored under the name `message`.
2. `print` is a software instruction requesting visible output.
3. The CPU ultimately executes lower-level operations that make that output appear.
