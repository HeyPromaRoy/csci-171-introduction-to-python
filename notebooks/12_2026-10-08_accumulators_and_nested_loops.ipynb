## Conceptual analogy

A cashier's running total is an accumulator. It starts once at zero, then each item's price is added. Resetting the total inside the item loop would erase earlier purchases. A nested loop is like processing every seat in every row: the outer loop chooses a row, and the inner loop visits seats in that row.

> This analogy provides an initial mental model. The definitions and Python examples below provide the precise technical meaning.


## Key terms

| Term | Plain-language meaning |
|---|---|
| **counter** | A variable tracking how many times or events occurred. |
| **accumulator** | A variable collecting a running result such as a sum. |
| **nested loop** | A loop inside another loop. |
| **outer loop** | The loop controlling larger groups. |
| **inner loop** | The loop completing repeated work inside each outer iteration. |


## Explanation


Initialize an accumulator once before the loop. Update it inside the loop with `total = total + value` or `total += value`. After the loop, the accumulator represents all processed values. For nested loops, keep examples small and trace ordered pairs such as `(row, seat)`.

## Predict before running

The next program intends to total three prices. Why will it print only the last price?

1. Predict the output or error.
2. Record the reason for the prediction.
3. Run the next cell only after recording both parts.


```python
prices = [3.0, 5.0, 2.0]
for price in prices:
    total = 0
    total = total + price
print("Total:", total)
```

### Result and diagnosis

`total` is reset to zero during every iteration. After the final iteration it contains only 2.0. The program runs, but the initialization is in the wrong location.

**Error-reading procedure:** Read the final line of the message, locate the referenced line, and compare the value or structure Python received with the program's intended behavior.


### Correction steps

1. Initialize the accumulator once before entering the loop.
2. Add each current value inside the loop.
3. Print the final accumulated result after the loop.


```python
prices = [3.0, 5.0, 2.0]
total = 0
for price in prices:
    total = total + price
    print("After adding", price, "the total is", total)
print("Final total:", total)
```

### How the corrected example works

1. The accumulator begins once at zero.
2. Each iteration adds exactly one price.
3. The final print is outside the loop because it reports the completed total.

A useful verification step is to change one value, predict the new result, and run the cell again.

### Practice 1: Track a count and total together

**Reference note:** A count records how many values were processed, while a total records their sum. The mean uses both after the loop.

**Before running:** Predict the exact output and identify the line responsible for each result.

**Lab change:** Change the range to summarize odd values from 1 through 9.


```python
total = 0
count = 0
for value in range(2, 11, 2):
    total = total + value
    count = count + 1

print("Count:", count)
print("Total:", total)
print("Mean:", total / count)
```

### Practice 2: Trace a small multiplication table

**Reference note:** The outer loop selects a base number and the inner loop selects its multiplier. Indentation shows which work repeats inside each outer iteration.

**Before running:** Predict the exact output and identify the line responsible for each result.

**Lab change:** Reduce both ranges, trace the first six outputs by hand, and then run the cell.


```python
for base in range(3, 7):
    print("Base:", base)
    for multiplier in range(1, 6):
        print(base, "x", multiplier, "=", base * multiplier)
```

### Practice 3: Build a rectangular text pattern

**Reference note:** A nested loop can separate rows from columns. The `end` argument keeps symbols on one row until the inner loop finishes.

**Before running:** Predict the exact output and identify the line responsible for each result.

**Lab change:** Create a rectangle with two rows and eight columns.


```python
rows = 3
columns = 5
for row in range(rows):
    for column in range(columns):
        print("*", end="")
    print()
```

### Practice 4: Count all nested-loop iterations

**Reference note:** The accumulator changes once for every inner-loop visit. Multiplying the two range lengths provides an independent check.

**Before running:** Predict the exact output and identify the line responsible for each result.

**Lab change:** Predict the count for five rows and two columns before changing the code.


```python
visits = 0
for row in range(3):
    for column in range(4):
        visits = visits + 1
        print("position", row, column)

print("Total positions:", visits)
```
