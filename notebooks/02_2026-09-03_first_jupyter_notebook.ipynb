## Conceptual analogy

A Jupyter notebook is like a science lab notebook with two kinds of pages. Markdown cells hold explanations and observations. Code cells hold experiments that Python can run. Like a real experiment, order matters: using a result before creating it is like trying to weigh a sample before placing it on the scale.

> This analogy provides an initial mental model. The definitions and Python examples below provide the precise technical meaning.


## Key terms

| Term | Plain-language meaning |
|---|---|
| **notebook** | A document containing executable code, explanations, and results. |
| **text cell** | Colab's name for a Markdown cell containing formatted notes, headings, lists, and explanations. |
| **code cell** | A cell containing Python instructions that can be executed. |
| **runtime (kernel)** | The temporary Python process that executes code and remembers the current session state. |
| **output** | The visible result produced when a code cell runs. |
| **NameError** | An error raised when Python cannot find a requested name. |

### Runtime state and reliable order

A code cell may create a value that the runtime remembers. A later cell can use that value only if the earlier definition has already run in the current session. Reopening the notebook preserves saved cells, but a new runtime does not automatically remember values from an earlier runtime.

The best reproducibility check is to use Colab's **Runtime** menu to restart the session and run all cells from the beginning. A notebook that works only after cells are run out of order is not reliable. Each name should be introduced before it is used.

### Result and diagnosis

Python raises `NameError: name 'course_name' is not defined`. The kernel has no memory of a value named `course_name` because the definition has not run.

**Error-reading procedure:** Read the final line of the message, locate the referenced line, and compare the value or structure Python received with the program's intended behavior.


### Correction steps

1. Create the name before the first place where it is used.
2. Run the definition cell.
3. Run the output cell, then restart the kernel and run all cells to confirm the order.


```python
course_name = "CSCI 171"
print(course_name)
print("Welcome to your first Python notebook!")
```

### How the corrected example works

1. The first line creates `course_name` and stores text in it.
2. The next line asks Python to display the stored text.
3. The final line prints a second string directly without storing it first.

A useful verification step is to change one value, predict the new result, and run the cell again.
