## Key terms

| Term | Plain-language meaning |
|---|---|
| **method** | A function attached to a value, called with dot notation: `value.method()`. |
| **return value** | The result a method hands back. For string methods, this is usually a new string, a number, or a Boolean. |
| **argument** | A value placed inside the parentheses to tell the method what to work with, such as the `"|"` in `split("|")`. |
| **normalize** | Convert equivalent inputs to a consistent form. |
| **whitespace** | Spaces, tabs (`\t`), and newlines (`\n`). |
| **delimiter** | The character or text that separates fields, such as a comma or `|`. |
| **list** | An ordered collection written in square brackets. `split` returns one; lists are studied in detail later. |

## Part 1: Changing case

Case methods produce a new string with different capitalization. They are the most common first step in normalization because Python treats `"Yes"`, `"YES"`, and `"yes"` as three different strings.

**Before running:** Predict each line, paying close attention to `title` and `swapcase`.


````python
course = "intro to PYTHON programming"

print("upper():     ", course.upper())
print("lower():     ", course.lower())
print("title():     ", course.title())
print("capitalize():", course.capitalize())
print("swapcase():  ", course.swapcase())
print("original:    ", course)
````

### What happened

```
upper():      INTRO TO PYTHON PROGRAMMING
lower():      intro to python programming
title():      Intro To Python Programming
capitalize(): Intro to python programming
swapcase():   INTRO TO python PROGRAMMING
original:     intro to PYTHON programming
```

- `upper` and `lower` change **every** letter. Characters that are not letters (spaces, digits, punctuation) pass through untouched.
- `title` capitalizes the first letter of **each word** and lowercases the rest of that word.
- `capitalize` capitalizes only the **first character of the whole string** and lowercases everything else — notice that `PYTHON` became `python`.
- `swapcase` flips each letter: uppercase becomes lowercase and lowercase becomes uppercase.
- The last line proves the original is unchanged: five methods ran, and `course` still holds the text it started with.

**Common use:** Compare user responses after normalizing: `answer.lower() == "yes"` accepts `Yes`, `YES`, and `yes`.

**Caution with `title`:** It starts a new "word" after any non-letter, so `"it's a dog's life".title()` returns `"It'S A Dog'S Life"`. Inspect `title` output before relying on it for names.


## Part 2: Removing characters at the edges

`strip` removes from **both** ends, `lstrip` from the **left** (start) only, and `rstrip` from the **right** (end) only. With no argument they remove whitespace. With an argument they remove any of the listed characters.

**Before running:** Predict the line that uses `strip("*")` directly on `raw_entry`. Will the stars disappear?


````python
raw_entry = "   ***Room 101***   "

print("strip():      [" + raw_entry.strip() + "]")
print("lstrip():     [" + raw_entry.lstrip() + "]")
print("rstrip():     [" + raw_entry.rstrip() + "]")
print("strip('*'):   [" + raw_entry.strip("*") + "]")

no_spaces = raw_entry.strip() #***Room 101***
no_stars = no_spaces.strip("*")
print("two steps:    [" + no_stars + "]")
````

### What happened

```
strip():      [***Room 101***]
lstrip():     [***Room 101***   ]
rstrip():     [   ***Room 101***]
strip('*'):   [   ***Room 101***   ]
two steps:    [Room 101]
```

- The square brackets make the edges visible, so leftover spaces can be seen.
- `strip("*")` on `raw_entry` removed **nothing**. Stripping works from the outside in and stops at the first character that is not in the removal set. The outermost characters are spaces, not stars, so it stopped immediately.
- The two-step version first removes the spaces, which exposes the stars at the edges; the second call can then remove them.
- None of these methods touch the inside of the string: the space in `Room 101` survives every call. To change characters in the middle, use `replace` (Part 5).

**Note:** The argument is a *set of characters*, not a word. `"--==title==--".strip("=-")` removes every `-` and `=` from both edges and returns `"title"`.


## Part 3: Searching inside a string

Searching tools answer three different questions:

- **Is it there?** → `in`, `startswith`, `endswith` return `True` or `False`.
- **Where is it?** → `find` and `rfind` return an index (or `-1`).
- **How many?** → `count` returns an integer.

**Before running:** Label the position of the `@` and each `.` in the email, then predict every line.


````python
email = "maya.lee@jjay.cuny.edu"
#                          18192021
#        0123456789101112131415161718.........21

print("'@' in email:       ", "@" in email)
print("Length: ", len(email))
print("'gmail' in email:   ", "gmail" in email)
print("find('@'):          ", email.find("@"))
print("find('.'):          ", email.find("."))
print("rfind('.'):         ", email.rfind("."))
print("find('gmail'):      ", email.find("gmail"))
print("count('.'):         ", email.count("."))
print("startswith('maya'): ", email.startswith("maya"))
print("endswith('.edu'):   ", email.endswith(".edu"))
````

### What happened

```
'@' in email:        True
'gmail' in email:    False
find('@'):           8
find('.'):           4
rfind('.'):          18
find('gmail'):       -1
count('.'):          3
startswith('maya'):  True
endswith('.edu'):    True
```

- `in` is an **operator**, not a method, so it is written `"@" in email`, without a dot or parentheses. It is the clearest way to ask a yes/no question.
- `find` scans from the **left** and reports the first match. The first `.` is at index 4 (between `maya` and `lee`).
- `rfind` scans from the **right** and reports the last match: the `.` before `edu` sits at index 18. The index itself is still counted from the left.
- When the text is absent, `find` returns `-1` instead of crashing.
- `count` reports how many times the text appears without overlapping.
- `startswith` and `endswith` check only the beginning or the end — useful for file extensions, prefixes, and domain checks.
- Searching is **case-sensitive**: `email.startswith("Maya")` would be `False`. Normalize with `lower()` first when case should not matter.


### Using `find` with slicing

Meeting 17 used a hard-coded `at_index = 8` to split an email. `find` locates the position automatically, so the same code works for any email.


````python
email = "maya.lee@jjay.cuny.edu"
at_position = email.find(".")

username = email[:at_position]
domain = email[at_position + 1:]

print("@ found at index:", at_position)
print("Username:", username)
print("Domain:", domain)
````

### What happened

```
@ found at index: 8
Username: maya.lee
Domain: jjay.cuny.edu
```

1. `find("@")` returned `8`, which was stored in `at_position`.
2. `email[:at_position]` takes everything **before** index 8.
3. `email[at_position + 1:]` starts one position **after** the `@`, so the `@` itself is excluded.

Changing the email to `"sam@cuny.edu"` now works without editing any number, because the position is computed rather than counted by hand.

All of them return `False` for an empty string `""`.

### Validating input before conversion

`int("abc")` raises `ValueError`. Testing with `isdigit` first lets the program respond politely. The value below stands in for text typed at an `input()` prompt.


````python
age_text = " 19 "            # stands in for input("Age: ")
cleaned_age = age_text.strip()

if cleaned_age.isdigit():
    age = int(cleaned_age)
    print("Next year you will be", age + 1)
else:
    print("Please enter a whole number.")
````

3. Change `age_text` to `"nineteen"` or `"-3"`, predict the output, and run again. Both reach the `else` branch.

The order *clean → test → convert* is a reusable pattern for any numeric input.


### Testing one character at a time

The `is` methods also work on single characters, which pairs well with a `for` loop and counters (accumulators).


````python
password = "Secure2026Pass!"
upper_count = 0
digit_count = 0
other_count = 0

for character in password:  # character = S e c ..... !
    if character.isupper():
        upper_count += 1 # 1
    elif character.isdigit():
        digit_count += 1
    elif not character.isalnum(): # A- Z , a -z , 0-9
        other_count += 1

print("Length:", len(password))
print("Uppercase letters:", upper_count)
print("Digits:", digit_count)
print("Symbols:", other_count)
````

### What happened

```
Length: 15
Uppercase letters: 2
Digits: 4
Symbols: 1
```

- `len(password)` is a built-in **function**, not a method, so it is written `len(password)` rather than `password.len()`.
- The loop sends one character at a time through the `if/elif` chain.
- `S` and `P` are uppercase, `2`, `0`, `2`, `6` are digits, and `!` fails `isalnum`, so it counts as a symbol. Lowercase letters match no branch and are not counted.
- This is the same accumulator pattern from earlier loops — only the conditions are new.


## Part 5: Replacing, splitting, and joining

- `replace(old, new)` substitutes text **anywhere** in the string, including the middle.
- `split(sep)` breaks a string into a list of pieces at each `sep`.
- `sep.join(pieces)` does the reverse: it glues pieces together with `sep` between them.

**Before running:** Predict what the optional third argument `1` does in the second `replace` call.


````python
note = "ha ha ha"

print(note.replace("ha", "HA"))
print(note.replace("ha", "HA", 5))
print(note.replace(" ", ""))
print(note.replace("xyz", ":"))
````

### What happened

```
HA HA HA
HA ha ha
hahaha
ha ha ha
```

- With two arguments, `replace` changes **every** match.
- A third argument limits the number of replacements, counted from the left — `1` changed only the first `ha`.
- Replacing `" "` with the empty string `""` deletes every space, including the inner ones that `strip` cannot reach.
- When the old text is not present, `replace` returns an unchanged copy instead of raising an error.


````python
raw_record = "Maya|CSCI 171|Fall 2026"
fields = raw_record.split("|")

print("fields:", fields)
print("number of fields:", len(fields))
print("first field:", fields[0])

joined = " / ".join(fields)
print("joined:", joined)

phrase = "learn   python    today"
print("split():   ", phrase.split())
print("split(' '):", phrase.split(" "))
````

### What happened

```
fields: ['Maya', 'CSCI 171', 'Fall 2026']
number of fields: 3
first field: Maya
joined: Maya / CSCI 171 / Fall 2026
split():    ['learn', 'python', 'today']
split(' '): ['learn', '', '', 'python', '', '', '', 'today']
```

- `split("|")` cut the record at each `|` and returned a **list** of three strings. The delimiter itself is removed.
- `len` works on lists too, and `fields[0]` uses the same zero-based indexing as strings.
- In `" / ".join(fields)`, the method is called on the **separator**, and the list goes in the parentheses. This reads as "join these fields using ` / `." Every item in the list must be a string.
- `split()` with **no argument** splits on any run of whitespace and drops the empty pieces — the best choice for splitting a sentence into words.
- `split(" ")` splits on **each single space**, so consecutive spaces produce empty strings `''`. Choose the delimiter deliberately.


## Part 6: Aligning output

Layout methods pad a string to a chosen **total width**:

- `ljust(w)` — text on the left, padding on the right
- `rjust(w)` — text on the right, padding on the left
- `center(w)` — text in the middle
- `zfill(w)` — pad on the left with zeros

An optional second argument to `ljust`, `rjust`, and `center` chooses the fill character.


````python
item = "Coffee"
price = "3.50"

print("[" + item.ljust(10) + "]")
print("[" + item.rjust(10) + "]")
print("[" + item.center(10, "*") + "]")
print("[" + price.zfill(6) + "]")
print("[" + item.ljust(3) + "]")

print()
print("RECEIPT".center(20, "="))
print("Tea".ljust(12) + "2.25".rjust(8))
print("Sandwich".ljust(12) + "7.80".rjust(8))
print("Coffee".ljust(12) + "3.50".rjust(8))
````

### What happened

```
[Coffee    ]
[    Coffee]
[**Coffee**]
[003.50]
[Coffee]

======RECEIPT=======
Tea             2.25
Sandwich        7.80
Coffee          3.50
```

- `"Coffee"` has 6 characters, so a width of 10 adds 4 padding characters.
- `center(10, "*")` split the 4 extra characters evenly, 2 on each side. With an odd amount, as in `"RECEIPT".center(20, "=")` (13 extra), one side receives one more.
- `zfill(6)` padded `"3.50"` with zeros to 6 characters — useful for IDs such as `"42".zfill(5)` → `"00042"`.
- When the width is **smaller** than the text, as in `ljust(3)`, the string is returned unchanged; it is never cut.
- In the receipt, every name is padded to 12 characters and every price to 8, so the columns line up regardless of word length.


## Part 7: Chaining order matters

Chained methods run **left to right**: each method operates on the string returned by the one before it. Changing the order can change the result.

**Before running:** Predict both lines. Which one produces underscores at the edges?


````python
raw_title = "  Intro To Python      lo"

strip_first = raw_title.strip().replace(" ", "_") # Intro_To_Python______lo
# otput for raw_title.strip() = Intro To Python output of .strip().replace = Intro_To_Python
replace_first = raw_title.replace("o", "_").strip().upper() #___Intro_To_PYthon___ #

print("strip then replace:", strip_first)
print("replace then strip:", replace_first)
````

### What happened

```
strip then replace: Intro_To_Python
replace then strip: __Intro_To_Python__
```

- **Strip first:** the edge spaces are removed, leaving only the inner spaces for `replace` to convert.
- **Replace first:** the edge spaces become underscores before `strip` runs. `strip` removes whitespace only, and underscores are not whitespace, so they remain.

**Readable order for cleaning:** strip the edges → normalize the case → replace or split. When a chain becomes hard to read, store each step in a named variable, as in the corrected example earlier.

The original still has its spaces (two after `Original:` plus the separator space from `print`). `strip` runs first, then `lower` works on the already-stripped string. For one-sided spaces in the lab change, `strip` still works; `lstrip` or `rstrip` would remove only one side.


### Practice 2: Replace a repeated separator

**Reference note:** `replace` creates a new string in which every matching piece is substituted. The source string remains unchanged.

**Before running:** Predict the exact output and identify the line responsible for each result.

**Lab change:** Replace the separator with a comma and one space.


````python
raw_record = "Maya|CSCI 171|Fall 2026"
readable_record = raw_record.replace("|", " a ")

print(raw_record)
print(readable_record)
````

**Output and explanation**

```
Maya|CSCI 171|Fall 2026
Maya - CSCI 171 - Fall 2026
```

Both `|` characters were replaced because `replace` changes every match by default. The same result is possible with `" - ".join(raw_record.split("|"))`; `replace` is simpler when the goal is only to swap one separator for another, while `split` is better when the separate fields are needed.


### Practice 3: Count a substring

**Reference note:** `count` reports non-overlapping occurrences. Normalizing case first prevents uppercase and lowercase versions from being counted separately.

**Before running:** Predict the exact output and identify the line responsible for each result.

**Lab change:** Add another occurrence with different capitalization and predict the new count.


````python
sentence = "Python practice makes Python clearer."
normalized = sentence.lower()

print("python count:", normalized.count("python"))
print("practice count:", normalized.count("practice"))
````

**Output and explanation**

```
python count: 2
practice count: 1
```

Without `lower`, `sentence.count("python")` would return `0`, because both occurrences start with a capital `P`. Non-overlapping means `"aaaa".count("aa")` is `2`, not `3`: after a match, the search continues after the end of that match.


### Practice 4: Clean a simple identifier

**Reference note:** Several small transformations can form a readable identifier. Each method has one visible purpose in the chain.

**Before running:** Predict the exact output and identify the line responsible for each result.

**Lab change:** Convert a different three-word title and inspect every transformation separately.


````python
raw_title = "  Intro To Python  "
identifier = raw_title.strip().lower().replace(" ", "_")

print("Identifier:", identifier)
print("Original unchanged:", raw_title)
````

**Output and explanation**

```
Identifier: intro_to_python
Original unchanged:   Intro To Python  
```

The chain runs strip → lower → replace. As Part 7 showed, `strip` must come before `replace`, or the edge spaces would become underscores. To inspect each transformation separately for the lab change, store each step in its own variable and print it.
