## Conceptual analogy

A function is like a labeled service window. The sign tells you its purpose, the required forms are its parameters, the documents you hand over are arguments, and the completed document returned to you is the return value.

> This analogy provides an initial mental model. The definitions and Python examples below provide the precise technical meaning.


## Key terms

| Term | Plain-language meaning |
|---|---|
| **definition** | The `def` block that describes a function's behavior. |
| **parameter** | A placeholder name in the function definition. |
| **argument** | A value supplied during a function call. |
| **return value** | The value sent back by `return`. |
| **docstring** | A short description placed inside a function. |


## Explanation


Define one clear job per function. Use a verb-based name such as `calculate_total`. Parameters make the function flexible. A function definition does not run the body; a call runs it. `return` immediately sends a value back, allowing later code to store, compare, print, or reuse that result.

### Correction steps

1. Keep the definition as the reusable recipe.
2. Add a call after the definition.

### Practice 2: Return the sum of three parameters

**Reference note:** Parameters receive the three arguments, and `return` sends the calculated sum back to the caller.

**Before running:** Predict the exact output and identify the line responsible for each result.

**Lab change:** Call the function with three decimal values and predict the result.


```python
def calculate_sum(number_1, number_2, number_3):
    return number_1 + number_2 + number_3

first_total = calculate_sum(1, 2, 3)
second_total = calculate_sum(10, -2, 5)
print(first_total)
print(second_total)
```

### Practice 3: Package temperature conversion in a function

**Reference note:** A parameter makes the earlier conversion example reusable. Returning the result allows later code to print, compare, or store it.

**Before running:** Predict the exact output and identify the line responsible for each result.

**Lab change:** Add a call for 100 degrees Celsius and verify the expected boiling point.


```python
def celsius_to_fahrenheit(celsius):
    fahrenheit = (celsius * 9 / 5) + 32
    return fahrenheit

print("Freezing:", celsius_to_fahrenheit(0))
print("Room temperature:", celsius_to_fahrenheit(20))
```

### Practice 4: Use a returned value in later work

**Reference note:** A returned value can be assigned to a name and used in another calculation. Printing inside a function alone would not provide that same flexibility.

**Before running:** Predict the exact output and identify the line responsible for each result.

**Lab change:** Test zero, decimal, and negative arguments, then explain which inputs make sense for the context.


```python
def rectangle_area(width, height):
    return width * height

classroom_area = rectangle_area(8, 6)
doubled_area = classroom_area * 2

print("One room:", classroom_area)
print("Two rooms:", doubled_area)
```
