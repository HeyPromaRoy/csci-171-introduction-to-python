## Predict before running

The next mixed program is supposed to add only positive values. Predict its result and find the logic problem.

1. Predict the output or error.
2. Record the reason for the prediction.
3. Run the next cell only after recording both parts.


```python
values = [3, -2, 5]
total = 0
for value in values:
    if value < 0:
        total += value
print(total)
```

### Result and diagnosis

The code adds only the negative value and prints -2. The comparison points in the opposite direction from the requirement.

**Error-reading procedure:** Read the final line of the message, locate the referenced line, and compare the value or structure Python received with the program's intended behavior.


### Correction steps

1. Underline the words `only positive values` in the requirement.
2. Translate that phrase into `value > 0`.
3. Trace all three values and compare the total with a hand calculation.


```python
values = [3, -2, 5]
total = 0
for value in values:
    if value > 0:
        total += value
print("Positive total:", total)
```

### How the corrected example works

1. The loop visits every list item.
2. Only values greater than zero reach the accumulator update.
3. The hand-calculated expected result is 3 + 5 = 8.

A useful verification step is to change one value, predict the new result, and run the cell again.

## Additional in-class code lab

### Practice 1: Review conversion, arithmetic, and decisions

**Reference note:** This mixed example follows a value from text conversion through arithmetic to a labeled decision. Each intermediate value can be checked separately.

**Before running:** Predict the exact output and identify the line responsible for each result.

**Lab change:** Test one value that finishes below 70 and one that finishes exactly at 70.


```python
raw_score = "76"
score = int(raw_score)
adjusted_score = score + 4

if adjusted_score >= 70:
    result = "meets the practice threshold"
else:
    result = "needs more review"

print("Adjusted score:", adjusted_score)
print("Result:", result)
```

### Practice 2: Review while-loop counting

**Reference note:** The loop classifies each generated value and keeps a count of those meeting the condition.

**Before running:** Predict the exact output and identify the line responsible for each result.

**Lab change:** Modify the condition to count multiples of three from 1 through 20.


```python
number = 1
even_count = 0
while number <= 10:
    if number % 2 == 0:
        even_count = even_count + 1
    number = number + 1

print("Even values from 1 through 10:", even_count)
```

### Practice 3: Review for-loop accumulation

**Reference note:** A condition inside a `for` loop selects which values contribute to the total. The printed label states exactly what was accumulated.

**Before running:** Predict the exact output and identify the line responsible for each result.

**Lab change:** Predict the total, verify it by hand, and then change the program to total even values.


```python
total = 0
for number in range(1, 11):
    if number % 2 == 1:
        total = total + number

print("Sum of odd values from 1 through 10:", total)
```

### Practice 4: Review a nested-loop trace

**Reference note:** A compact nested loop tests whether the outer and inner control variables are understood before the assessment.

**Before running:** Predict the exact output and identify the line responsible for each result.

**Lab change:** Write the six coordinate pairs before running, then explain why the total is six.


```python
output_count = 0
for outer in range(2):
    for inner in range(3):
        print(outer, inner)
        output_count = output_count + 1

print("Lines produced:", output_count)
```
