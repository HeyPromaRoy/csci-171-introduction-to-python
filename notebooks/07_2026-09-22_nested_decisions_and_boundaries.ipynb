## Conceptual analogy

At an airport, the first checkpoint asks whether you have a boarding pass. Only after the answer is yes does a second checkpoint ask whether the name matches the ID. That second question is nested because it makes sense only inside the first successful path.

> This analogy provides an initial mental model. The definitions and Python examples below provide the precise technical meaning.


## Key terms

| Term | Plain-language meaning |
|---|---|
| **nested decision** | An `if` statement inside another decision block. |
| **range** | A set of values between boundaries. |
| **edge case** | A valid but unusual or boundary input. |
| **test case** | A chosen input and expected result used to check behavior. |
| **unreachable branch** | A branch that can never run because an earlier condition always captures it. |


## Explanation


Use nesting when the second question depends on the first. Do not nest simply because code looks “more advanced.” For ranges, condition order is crucial. If `score >= 70` appears before `score >= 90`, a score of 95 matches the first condition and never reaches the excellent branch.

Build a test table before coding. For boundaries 70 and 90, useful tests include 69, 70, 71, 89, 90, and 91.

## Predict before running

The next code orders broad and narrow conditions incorrectly. Predict the message for a score of 95.

1. Predict the output or error.
2. Record the reason for the prediction.
3. Run the next cell only after recording both parts.


```python
score = 95

if score >= 70:
    print("Passing")
elif score >= 90:
    print("Excellent")
else:
    print("Needs support")
```

### Result and diagnosis

The code prints `Passing`. This is a logic error, not a syntax error. Because 95 is also at least 70, Python takes the first true branch and never checks `score >= 90`.

**Error-reading procedure:** Read the final line of the message, locate the referenced line, and compare the value or structure Python received with the program's intended behavior.


### Correction steps

1. Identify conditions that overlap.
2. Place the most restrictive or highest threshold first.
3. Run test cases on both sides of every boundary.


```python
score = 95

if score >= 90:
    message = "Excellent"
elif score >= 70:
    message = "Passing"
else:
    message = "Needs support"

print(score, "->", message)
```

### How the corrected example works

1. The highest threshold is checked first.
2. The second branch receives values from 70 through 89.
3. The same code should be rerun with 69, 70, 89, 90, and 95 to check the boundaries.

A useful verification step is to change one value, predict the new result, and run the cell again.

**Reference note:** The inner ID check matters only after the age requirement succeeds. Indentation makes that dependency visible.

### Practice 2: Classify a boundary value

**Reference note:** Exactly one branch should describe the number. Zero needs its own equality branch because it is neither positive nor negative.

**Before running:** Predict the exact output and identify the line responsible for each result.

**Lab change:** Test `-1`, `0`, and `1`, then explain why the equality check belongs before `else`.


```python
number = 0
if number > 0:
    print("positive")
elif number == 0:
    print("zero")
else:
    print("negative")
```

### Practice 3: Apply a nested shipping rule

**Reference note:** The outer branch separates local from nonlocal orders. Only a local order needs the inner weight decision.

**Before running:** Predict the exact output and identify the line responsible for each result.

**Lab change:** Test weights 10 and 11 for both local and nonlocal delivery.


```python
local_delivery = True
weight = 8

if local_delivery:
    if weight <= 10:
        shipping = 5
    else:
        shipping = 8
else:
    shipping = 12

print("Shipping charge:", shipping)
```

### Practice 4: Combine score and attendance boundaries

**Reference note:** A course rule may require two independent thresholds. Testing the values exactly at each threshold checks whether the rule is inclusive.

**Before running:** Predict the exact output and identify the line responsible for each result.

**Lab change:** Test values one point below and exactly at both thresholds.


```python
score = 70
attendance = 80

if score >= 70:
    if attendance >= 80:
        result = "requirements met"
    else:
        result = "attendance requirement not met"
else:
    result = "score requirement not met"

print(result)
```
