## Conceptual analogy

A `for` loop is like checking each name on an attendance list. The list already determines how many names exist, so you do not manage a separate counter just to know when to stop. `range` creates an orderly sequence of numbers for Python to visit.

> This analogy provides an initial mental model. The definitions and Python examples below provide the precise technical meaning.


## Key terms

| Term | Plain-language meaning |
|---|---|
| **iterable** | A value whose items can be visited one at a time. |
| **loop variable** | The name receiving the current item during each iteration. |
| **`range`** | A sequence of integers ending before its stop value. |
| **step** | The amount added between range values. |
| **off-by-one error** | A boundary mistake that repeats once too few or too many times. |


## Explanation


`range(stop)` starts at 0. `range(start, stop)` uses a chosen start. `range(start, stop, step)` also controls movement. The stop value is excluded. To include 5 while counting upward from 1, use `range(1, 6)`.

Prefer `for` when iterating through a known collection or number sequence. Prefer `while` when repetition depends on a changing condition whose iteration count is not known in advance.

## Predict before running

The next code is intended to print 1 through 5. Which value will be missing?

1. Predict the output or error.
2. Record the reason for the prediction.
3. Run the next cell only after recording both parts.


```python
for number in range(1, 5):
    print(number)
```

### Result and diagnosis

The output stops at 4 because the range stop value is excluded. This is a classic off-by-one logic error.

**Error-reading procedure:** Read the final line of the message, locate the referenced line, and compare the value or structure Python received with the program's intended behavior.


### Correction steps

1. State the first and last desired values.
2. For an upward range, set stop to one past the desired last value.
3. List the expected sequence before running the loop.


```python
for number in range(1, 6):
    print(number)
```

### How the corrected example works

1. The first value is 1.
2. The stop value 6 is excluded.
3. Therefore the visited values are 1, 2, 3, 4, and 5.

A useful verification step is to change one value, predict the new result, and run the cell again.

### Practice 3: Iterate through text

### Practice 4: Accumulate values produced by range

**Reference note:** The loop variable supplies one number per iteration, and the accumulator preserves the running sum.

**Before running:** Predict the exact output and identify the line responsible for each result.

**Lab change:** Change the range so the program totals the multiples of 5 from 5 through 50.


```python
total = 0
for number in range(1, 6):
    total = total + number
    print("After", number, "the total is", total)

print("Final total:", total)
```
