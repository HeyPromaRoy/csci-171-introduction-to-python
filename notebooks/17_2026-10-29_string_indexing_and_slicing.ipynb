## Conceptual analogy

A string is like a row of numbered train cars. The first car is numbered 0 in Python. Negative indexes count backward from the final car. A slice chooses a connected group of cars, but the stop index marks the first car not included.

> This analogy provides an initial mental model. The definitions and Python examples below provide the precise technical meaning.


## Key terms

| Term | Plain-language meaning |
|---|---|
| **character** | One item in a string. |
| **index** | An integer position, starting at 0 from the left. |
| **negative index** | A position counted from the right, starting at -1. |
| **slice** | A selected range written as `start:stop:step`. |
| **immutable** | Unable to be changed in place after creation. |


## Explanation


For `word = "Python"`, valid positive indexes are 0 through 5. `word[0]` is `P`; `word[-1]` is `n`. In `word[1:4]`, positions 1, 2, and 3 are included while 4 is excluded. Strings are immutable, so build a new string instead of assigning to one character.

```python
word = "Python"
print(word[6])
```

### Result and diagnosis

Python raises `IndexError` because the six characters occupy indexes 0 through 5. The length is 6, but 6 is not a valid position.

**Error-reading procedure:** Read the final line of the message, locate the referenced line, and compare the value or structure Python received with the program's intended behavior.


### Correction steps

1. Use `len(word)` to count characters.
2. Remember that the last positive index is `len(word) - 1`.
3. Use `-1` when the last character is the intended target.


```python
word = "Python"
print("Length:", len(word))
print("First:", word[0])
print("Last:", word[-1])
print("Middle slice:", word[1:4])
print("Reversed:", word[::-1])
```

### How the corrected example works

1. Length counts characters, while indexes begin at zero.
2. A negative index is often safer for the final character.
3. A step of -1 traverses the string backward.

A useful verification step is to change one value, predict the new result, and run the cell again.

### Practice 2: Slice a string

**Reference note:** A slice includes the start index and excludes the stop index. Omitting one endpoint means continue to that end of the string.

**Before running:** Predict the exact output and identify the line responsible for each result.

**Lab change:** Label the indexes on paper and predict each slice.


```python
message = "Introduction"
print(message[0:5])
print(message[:5])
print(message[5:])
print(message[::2])
```

### Practice 3: Reverse text with a negative step

**Reference note:** A step of minus one walks from the end toward the beginning. The original string is unchanged because strings are immutable.

**Before running:** Predict the exact output and identify the line responsible for each result.

**Lab change:** Use the same pattern to test whether a short word reads the same backward.


```python
word = "notebook"
reversed_word = word[::-1]

print("Original:", word)
print("Reversed:", reversed_word)
```

### Practice 4: Loop through indexes

**Reference note:** `range(len(text))` provides every valid index. Keeping both the index and character visible supports careful tracing.

**Before running:** Predict the exact output and identify the line responsible for each result.

**Lab change:** Change the text, predict the final valid index, and verify it.


```python
text = "code"
for index in range(len(text)):
    print(index, "->", text[index])
```
