## Conceptual analogy

A form at a front desk receives everything as writing, even when someone writes a number. The staff member must interpret the writing before calculating with it. In the same way, `input` gives Python a string. Conversion is the step that says, “treat this text as a number.”

> This analogy provides an initial mental model. The definitions and Python examples below provide the precise technical meaning.


## Key terms

| Term | Plain-language meaning |
|---|---|
| **input** | Data supplied to a program. |
| **type conversion** | Creating a value of one type from a compatible value of another type. |
| **operator** | A symbol such as `+`, `-`, `*`, `/`, `//`, `%`, or `**`. |
| **expression** | Values and operators that Python evaluates to produce a result. |
| **concatenation** | Joining strings with `+`. |
| **modulo (`%`)** | The remainder after division. |


## Explanation


`input("Age: ")` pauses and returns text. If the user types `20`, Python receives `"20"`. Use `int(...)` for whole numbers or `float(...)` for decimals. Parentheses make the intended order clear. The arithmetic operators `/`, `//`, and `%` answer different questions: exact division, whole-number quotient, and remainder.

First inspect the data, then convert it, then calculate. Combining all three steps immediately can make an error harder to locate.

```python
quantity_text = "3"
total_items = quantity_text + 2
print(total_items)
```

### Result and diagnosis

Python raises a `TypeError` because string concatenation and numeric addition are different operations. Python will not guess which one you intended.

**Error-reading procedure:** Read the final line of the message, locate the referenced line, and compare the value or structure Python received with the program's intended behavior.


### Correction steps

1. Inspect the input value and confirm that its type is `str`.
2. Convert the numeric-looking text with `int`.
3. Perform the calculation and print a labeled result.


```python
quantity_text = "3"
quantity = int(quantity_text)
total_items = quantity + 2
print("Total items:", total_items)
```

### How the corrected example works

1. `quantity_text` preserves the original text.
2. `int(quantity_text)` creates the number 3.
3. Only then does `+ 2` perform numeric addition.

A useful verification step is to change one value, predict the new result, and run the cell again.

## Additional in-class code lab

### Practice 1: Convert numeric-looking text

**Reference note:** Input arrives as text even when it contains digits. Convert before arithmetic, and keep the original text available when comparing types.

**Before running:** Predict the exact output and identify the line responsible for each result.

**Lab change:** Change the text to another whole number and predict all three output lines.


```python
raw_age = "19"
age = int(raw_age)

print(raw_age, type(raw_age))
print(age, type(age))
print("Age next year:", age + 1)
```

### Practice 2: Convert Celsius to Fahrenheit

**Reference note:** The conversion multiplies Celsius by nine-fifths and then adds 32. Parentheses make the intended arithmetic order easy to verify.

**Before running:** Predict the exact output and identify the line responsible for each result.

**Lab change:** Test the freezing point of water and verify the result by hand.


```python
celsius = 20.0
fahrenheit = (celsius * 9 / 5) + 32

print("Celsius:", celsius)
print("Fahrenheit:", fahrenheit)
```

### Practice 3: Compare division, floor division, and remainder

**Reference note:** Floor division and remainder separate complete groups from leftovers. Ordinary division answers a different question.

**Before running:** Predict the exact output and identify the line responsible for each result.

**Lab change:** Repeat with 23 items in groups of 4 and check that groups times size plus remainder equals the original count.


```python
items = 17
group_size = 5

print("Ordinary division:", items / group_size)
print("Complete groups:", items // group_size)
print("Left over:", items % group_size)
```

### Practice 4: Calculate and label a purchase total

**Reference note:** Arithmetic is easier to inspect when intermediate values have meaningful names. A labeled output also communicates what the number represents.

**Before running:** Predict the exact output and identify the line responsible for each result.

**Lab change:** Add a fixed discount before tax and explain where it belongs in the sequence.


```python
price = 3.75
quantity = 4
subtotal = price * quantity
tax_rate = 0.08875
total = subtotal + subtotal * tax_rate

print("Subtotal:", subtotal)
print("Total:", round(total, 2))
```
