## Conceptual analogy

A form clerk does not simply accept an impossible birth month. The clerk explains the valid range and asks again. Validation is not punishment; it protects the quality of later work by ensuring the data meets a known rule.

> This analogy provides an initial mental model. The definitions and Python examples below provide the precise technical meaning.


## Key terms

| Term | Plain-language meaning |
|---|---|
| **validation** | Checking whether input satisfies required rules. |
| **sentinel** | A special value that signals stop or exit. |
| **valid range** | The accepted minimum and maximum values. |
| **defensive programming** | Anticipating and handling predictable input problems. |


## Explanation


A validation loop begins with a value, checks whether it is invalid, explains the rule, and obtains a new value inside the loop. The prompt must appear again inside the loop; otherwise the invalid value never changes. A sentinel such as `-1` can mean “stop,” but it should not be confused with ordinary data.

## Predict before running

The next loop detects an invalid value but never obtains a replacement. What will happen?

1. Predict the output or error.
2. Record the reason for the prediction.
3. Run the next cell only after recording both parts.


```python
month = 15
while month < 1 or month > 12:
    print("Month must be 1 through 12")
    # month is never replaced
```

### Result and diagnosis

The invalid value stays 15, so the condition remains true. This is another infinite-loop pattern: the message repeats, but the data never changes.

**Error-reading procedure:** Read the final line of the message, locate the referenced line, and compare the value or structure Python received with the program's intended behavior.


### Correction steps

1. Write the valid range in plain language.
2. Express the invalid condition with `or`.
3. Update the value inside the loop.
4. Test the lower boundary, upper boundary, and one value outside each boundary.


```python
month = 15
attempt_number = 1

while month < 1 or month > 12:
    print(month, "is invalid; month must be 1 through 12")
    if attempt_number == 1:
        month = 0
    else:
        month = 7
    attempt_number = attempt_number + 1

print(month, "is valid")
```

### How the corrected example works

1. The example begins with invalid value 15 and uses 0 as the next simulated attempt.
2. The second replacement is 7, so the value eventually satisfies the rule.
3. The loop stops when the value 7 satisfies the range rule.

A useful verification step is to change one value, predict the new result, and run the cell again.

## Additional in-class code lab

### Practice 1: Move an invalid value into range

**Reference note:** A validation loop repeats while the state violates the rule. This non-interactive model changes the value explicitly so the trace is safe to run.

**Before running:** Predict the exact output and identify the line responsible for each result.

**Lab change:** Start at a different invalid value and choose an update that eventually reaches 1 through 5.


```python
rating = -2
while rating < 1 or rating > 5:
    print("Rejected rating:", rating)
    rating = rating + 3

print("Accepted rating:", rating)
```

### Practice 2: Validate a menu choice

**Reference note:** The loop rejects anything outside the allowed text choices. The example simulates a corrected second attempt without pausing for keyboard input.

**Before running:** Predict the exact output and identify the line responsible for each result.

**Lab change:** Change the simulated correction to each valid menu choice and verify that the loop stops.


```python
choice = "9"
attempts = 0
while choice != "1" and choice != "2" and choice != "3":
    print("Invalid choice:", choice)
    attempts = attempts + 1
    choice = "2"

print("Accepted choice:", choice)
print("Rejected attempts:", attempts)
```

### Practice 3: Protect division from a zero denominator

**Reference note:** The program must establish a nonzero denominator before division. Validation belongs before the risky arithmetic operation.

**Before running:** Predict the exact output and identify the line responsible for each result.

**Lab change:** Use a negative nonzero denominator and explain why it is valid for division.


```python
denominator = 0
while denominator == 0:
    print("Zero cannot be used as the denominator.")
    denominator = 4

result = 20 / denominator
print("Result:", result)
```

### Practice 4: Limit repeated attempts

**Reference note:** A validation loop can also stop after a maximum number of attempts. Both parts of the `and` condition must remain true for another iteration.

**Before running:** Predict the exact output and identify the line responsible for each result.

**Lab change:** Change the correction point so the example reaches the attempt limit instead.


```python
access_code = "wrong"
attempts = 0
maximum_attempts = 3

while access_code != "python" and attempts < maximum_attempts:
    attempts = attempts + 1
    print("Attempt", attempts, "was not accepted.")
    if attempts == 2:
        access_code = "python"

print("Access granted:", access_code == "python")
```
