## Conceptual analogy

Cleaning text is like preparing names for a guest list. `strip` removes accidental space at the edges, `lower` standardizes capitalization for comparison, and `split` separates a line into meaningful pieces. The original sign-in sheet remains unchanged unless you store the cleaned version.

> This analogy provides an initial mental model. The definitions and Python examples below provide the precise technical meaning.


## Key terms

| Term | Plain-language meaning |
|---|---|
| **method** | A function attached to a value, called with dot notation. |
| **normalize** | Convert equivalent inputs to a consistent form. |
| **`strip`** | Return text without leading or trailing whitespace. |
| **`split`** | Return pieces separated by whitespace or a chosen delimiter. |
| **`find`** | Return the first index of a substring or -1 when absent. |


## Explanation


String methods do not modify the original string. `cleaned = raw.strip().lower()` stores a new result. Normalize before comparing user-entered text. Split only after deciding what separates fields. Print intermediate results while learning so each transformation is visible.

## Predict before running

The next code calls `strip` but ignores its returned value. Will `raw_name` change?

1. Predict the output or error.
2. Record the reason for the prediction.
3. Run the next cell only after recording both parts.


```python
raw_name = "  Proma Roy  "
raw_name.strip()
print("[" + raw_name + "]")
```

### Result and diagnosis

The spaces remain because strings are immutable and the new stripped string was not stored.

**Error-reading procedure:** Read the final line of the message, locate the referenced line, and compare the value or structure Python received with the program's intended behavior.


### Correction steps

1. Call the method on the original string.
2. Assign the returned string to a new descriptive name.
3. Print markers around the result to make edge spaces visible.


```python
raw_name = "  Proma Roy  "
clean_name = raw_name.strip()
normalized_name = clean_name.lower()
print("Raw:       [" + raw_name + "]")
print("Clean:     [" + clean_name + "]")
print("Normalized:", normalized_name)
```

### How the corrected example works

1. The raw input is preserved for comparison.
2. `strip` returns a new string without edge whitespace.
3. `lower` creates a consistent version for comparison.

A useful verification step is to change one value, predict the new result, and run the cell again.

## Additional in-class code lab

### Practice 1: Normalize surrounding space and case

**Reference note:** String methods return new strings. Chaining `strip` and `lower` creates a normalized value while preserving the original.

**Before running:** Predict the exact output and identify the line responsible for each result.

**Lab change:** Try text with mixed capitalization and extra spaces on only one side.


```python
raw_text = "  Python Practice  "
cleaned_text = raw_text.strip().lower()

print("Original:", raw_text)
print("Cleaned:", cleaned_text)
```

### Practice 2: Replace a repeated separator

**Reference note:** `replace` creates a new string in which every matching piece is substituted. The source string remains unchanged.

**Before running:** Predict the exact output and identify the line responsible for each result.

**Lab change:** Replace the separator with a comma and one space.


```python
raw_record = "Maya|CSCI 171|Fall 2026"
readable_record = raw_record.replace("|", " - ")

print(raw_record)
print(readable_record)
```

### Practice 3: Count a substring

**Reference note:** `count` reports non-overlapping occurrences. Normalizing case first prevents uppercase and lowercase versions from being counted separately.

**Before running:** Predict the exact output and identify the line responsible for each result.

**Lab change:** Add another occurrence with different capitalization and predict the new count.


```python
sentence = "Python practice makes Python clearer."
normalized = sentence.lower()

print("python count:", normalized.count("python"))
print("practice count:", normalized.count("practice"))
```

### Practice 4: Clean a simple identifier

**Reference note:** Several small transformations can form a readable identifier. Each method has one visible purpose in the chain.

**Before running:** Predict the exact output and identify the line responsible for each result.

**Lab change:** Convert a different three-word title and inspect every transformation separately.


```python
raw_title = "  Intro To Python  "
identifier = raw_title.strip().lower().replace(" ", "_")

print("Identifier:", identifier)
print("Original unchanged:", raw_title)
```
