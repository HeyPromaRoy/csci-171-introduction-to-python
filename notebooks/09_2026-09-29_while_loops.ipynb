## Conceptual analogy

Waiting for an elevator is a `while` loop: while the elevator has not arrived, keep watching the indicator. The condition must eventually change, or you would wait forever. In code, the update is the event that moves the loop toward stopping.

> This analogy provides an initial mental model. The definitions and Python examples below provide the precise technical meaning.


## Key terms

| Term | Plain-language meaning |
|---|---|
| **iteration** | One complete pass through a loop body. |
| **loop condition** | The Boolean checked before each iteration. |
| **initialization** | The starting value created before a loop. |
| **update** | A change that moves the loop toward completion. |
| **infinite loop** | A loop whose condition never becomes false. |


## Explanation


A reliable counter-controlled loop has four visible parts: initialize a counter, test a condition, perform the repeated work, and update the counter. Trace the value before and after each iteration. Ask: What is true before the loop? Why should it eventually stop? What is true after it stops?

## Predict before running

The next cell is an infinite-loop pattern. Do not run it without using the notebook Stop button. Identify the missing part first.

1. Predict the output or error.
2. Record the reason for the prediction.
3. Run the next cell only after recording both parts.


```python
count = 1
while count <= 5:
    print(count)
    # count never changes
```

### Result and diagnosis

The condition stays true because `count` remains 1. If run, the cell prints 1 repeatedly until interrupted. The missing update prevents progress.

**Error-reading procedure:** Read the final line of the message, locate the referenced line, and compare the value or structure Python received with the program's intended behavior.


### Correction steps

1. State the starting and stopping values.
2. Add an update inside the loop body.
3. Trace the counter through every iteration.


```python
count = 1
while count <= 5:
    print(count)
    count = count + 1

print("Loop finished with count =", count)
```

### How the corrected example works

1. `count` begins at 1.
2. The body runs while `count <= 5`.
3. The update increases `count`; after printing 5, it becomes 6 and the loop stops.

A useful verification step is to change one value, predict the new result, and run the cell again.

### Practice 3: Build a countdown

**Reference note:** A decreasing loop needs an update that moves the state toward the stopping condition. The final message appears after the loop ends.

**Before running:** Predict the exact output and identify the line responsible for each result.

### Practice 4: Accumulate a running total

**Reference note:** The accumulator begins at the identity value zero and changes once per iteration. A separate counter controls how many values are added.

**Before running:** Predict the exact output and identify the line responsible for each result.

**Lab change:** Change the update so the loop totals only odd values from 1 through 9.


```python
number = 1
total = 0
while number <= 5:
    total = total + number
    print("Added", number, "running total", total)
    number = number + 1

print("Final total:", total)
```
