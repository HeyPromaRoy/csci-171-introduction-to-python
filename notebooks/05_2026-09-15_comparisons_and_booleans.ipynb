## Conceptual analogy

A building entrance may ask several yes/no questions: Is the ID valid? Is the person on the list? Is the building open? Each question produces a Boolean answer. `and` requires all gates to open, while `or` needs at least one acceptable path.

> This analogy provides an initial mental model. The definitions and Python examples below provide the precise technical meaning.


## Key terms

| Term | Plain-language meaning |
|---|---|
| **Boolean** | One of the values `True` or `False`. |
| **comparison** | An expression comparing two values. |
| **equality (`==`)** | A test asking whether two values are equal. |
| **logical operator** | `and`, `or`, or `not`, used to combine or reverse conditions. |
| **boundary** | An edge value where a condition changes from false to true or vice versa. |


## Explanation


Comparisons do not print a judgment automatically; they *produce* a Boolean value. Examples include `age >= 18`, `name == "Mina"`, and `score != 0`. Use `and` when every condition must be true. Use `or` when either condition is enough. Use `not` to reverse a Boolean.

Always test boundary values. For `age >= 18`, try 17, 18, and 19. The exact boundary often reveals whether `>` or `>=` is correct.


### Truth table for `and` and `or`

The table shows every possible combination of two Boolean values:

| A | B | A and B | A or B |
|---|---|---|---|
| `False` | `False` | `False` | `False` |
| `False` | `True` | `False` | `True` |
| `True` | `False` | `False` | `True` |
| `True` | `True` | `True` | `True` |

Notice that `and` is `True` only when both values are `True`. `or` is `True` when at least one value is `True`.

## Predict before running

The next cell uses `=` where a comparison is needed. Will Python treat it as a question?

1. Predict the output or error.
2. Record the reason for the prediction.
3. Run the next cell only after recording both parts.


```python
age = 18
if age = 18:
    print("Age is exactly 18")
```

### Result and diagnosis

Python raises a `SyntaxError` because `=` performs assignment and cannot be used as the equality test in this condition. The equality operator is `==`.

**Error-reading procedure:** Read the final line of the message, locate the referenced line, and compare the value or structure Python received with the program's intended behavior.


### Correction steps

1. Use `=` only to assign a value to a name.
2. Use `==` to ask whether two values are equal.
3. Test one value below, at, and above the boundary.


```python
age = 18
is_exactly_eighteen = age == 18
is_adult = age >= 18

print("Exactly 18?", is_exactly_eighteen)
print("Adult?", is_adult)
```

### How the corrected example works

1. `age = 18` stores a value.
2. `age == 18` asks a question and produces a Boolean.
3. `age >= 18` includes 18 and every greater value.

A useful verification step is to change one value, predict the new result, and run the cell again.

## Additional in-class code lab

### Practice 1: Evaluate relational operators

**Reference note:** A comparison produces `True` or `False`. Reading several related comparisons helps distinguish equality from ordering.

**Before running:** Predict the exact output and identify the line responsible for each result.

**Lab change:** Change `value` to 8 and predict each Boolean before running.


```python
value = 5
print("Equal to 5:", value == 5)
print("Not equal to 5:", value != 5)
print("Less than 8:", value < 8)
print("At least 5:", value >= 5)
```

### Practice 2: Combine conditions with and

**Reference note:** An `and` expression is true only when both required conditions are true. Meaningful names make the combined rule easier to audit.

**Before running:** Predict the exact output and identify the line responsible for each result.

**Lab change:** Test one case with sufficient age but no ID and one case with ID but insufficient age.


```python
age = 20
has_id = True
can_enter = age >= 18 and has_id

print("Meets age requirement:", age >= 18)
print("Has ID:", has_id)
print("Can enter:", can_enter)
```

### Practice 3: Combine alternatives with or

**Reference note:** An `or` expression is true when at least one acceptable route is true. This differs from a rule requiring every condition.

**Before running:** Predict the exact output and identify the line responsible for each result.

**Lab change:** Test all four combinations of the two Boolean values.


```python
has_student_pass = False
has_day_ticket = True
may_board = has_student_pass or has_day_ticket

print("May board:", may_board)
print("Has neither option:", not may_board)
```

### Practice 4: Test inclusive boundaries

**Reference note:** Boundary values reveal whether a rule should use `<` or `<=`. This example includes both endpoints of the accepted interval.

**Before running:** Predict the exact output and identify the line responsible for each result.

**Lab change:** Test scores `-1`, `0`, `69`, `70`, `100`, and `101` one at a time.


```python
score = 70
in_valid_range = 0 <= score <= 100
passing = score >= 70

print("Valid score:", in_valid_range)
print("Passing score:", passing)
```
