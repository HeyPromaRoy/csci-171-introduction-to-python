## Conceptual analogy

A variable is like a labeled shelf location, not a permanently sealed box. The label helps us find the current value, and the value can be replaced later. The label `score` may point to 80 now and 85 after a correction. Python tracks the type of the value currently stored there.

> This analogy provides an initial mental model. The definitions and Python examples below provide the precise technical meaning.


## Key terms

| Term | Plain-language meaning |
|---|---|
| **variable** | A name that refers to a value. |
| **assignment** | The act of binding a name to a value with `=`. |
| **integer (`int`)** | A whole number such as 7 or -2. |
| **float** | A number with a decimal component such as 3.5. |
| **string (`str`)** | Text enclosed in quotation marks. |
| **Boolean (`bool`)** | The value `True` or `False`. |


## Explanation


Read `score = 80` as “assign the value 80 to the name score,” not as an algebra equation. Python names may use letters, digits, and underscores, but may not begin with a digit. Quotation marks matter: `25` is a number, while `"25"` is text. Use `type(value)` when you need evidence.

Reassignment happens from right to left. In `score = score + 5`, Python first reads the old value, adds 5, and then stores the new result under the same name.

### Result and diagnosis

Python raises a `SyntaxError`. A name cannot begin with a digit because Python would not be able to read it as an identifier.

**Error-reading procedure:** Read the final line of the message, locate the referenced line, and compare the value or structure Python received with the program's intended behavior.


### Correction steps

1. Choose a descriptive name that begins with a letter or underscore.
2. Use lowercase words separated by underscores.
3. Inspect the value and its type after assignment.


```python
student_count = 25
course_name = "Introduction to Python"
average_score = 88.5
class_is_in_person = True

print(student_count, type(student_count))
print(course_name, type(course_name))
print(average_score, type(average_score))
print(class_is_in_person, type(class_is_in_person))
```

### How the corrected example works

1. Each name describes what its value represents.
2. Quotation marks create text; no quotation marks are used for numbers or Booleans.
3. `type` reports the actual current type.

A useful verification step is to change one value, predict the new result, and run the cell again.
