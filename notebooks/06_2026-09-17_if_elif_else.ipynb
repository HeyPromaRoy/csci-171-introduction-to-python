## Conceptual analogy

A transit sign sends travelers down different paths: if the downtown train is running, take it; elif a shuttle is available, use the shuttle; else, choose another route. Python checks paths from top to bottom and takes the first path whose condition is true.

> This analogy provides an initial mental model. The definitions and Python examples below provide the precise technical meaning.


## Key terms

| Term | Plain-language meaning |
|---|---|
| **branch** | One possible path through a decision. |
| **condition** | A Boolean expression controlling whether a branch runs. |
| **indentation** | Leading spaces that group statements into a block. |
| **`elif`** | An additional condition checked only if earlier conditions were false. |
| **`else`** | The fallback branch when no earlier condition is true. |


## Explanation


An `if` statement ends its condition with a colon. The indented lines belong to that branch. `elif` adds another question, while `else` handles everything remaining. Only one branch in a connected `if`/`elif`/`else` chain runs.

## Predict before running

The next cell forgets to indent the line controlled by `if`. Predict what Python will report.

1. Predict the output or error.
2. Record the reason for the prediction.
3. Run the next cell only after recording both parts.


```python
temperature = 85
if temperature > 80:
print("It is warm")
```

### Result and diagnosis

Python raises an `IndentationError` because it expects an indented block after the `if` line.

**Error-reading procedure:** Read the final line of the message, locate the referenced line, and compare the value or structure Python received with the program's intended behavior.


### Correction steps

1. Keep the colon at the end of the `if` condition.
2. Indent every statement that belongs to that branch by four spaces.
3. Add other branches only when the requirement needs them.


```python
temperature = 85

if temperature > 80:
    print("It is warm")
elif temperature >= 60:
    print("It is mild")
else:
    print("It is cool")
```

### How the corrected example works

1. Python checks `temperature > 80` first.
2. Because 85 satisfies that condition, the first message prints.
3. The remaining branches are skipped.

A useful verification step is to change one value, predict the new result, and run the cell again.


## Common bug: repeated `if` instead of `elif`

The next cell uses three separate `if` statements instead of an `elif` chain. Each `if` is checked independently, so more than one branch can run for the same score.

1. Predict how many lines will print for `score = 95`.
2. Record the reason for the prediction.
3. Run the next cell only after recording both parts.

```python
score = 95

if score >= 90:
    print("Excellent")
if score >= 70:
    print("Passing")
if score >= 0:
    print("Needs support")
```

### Result and diagnosis

All three messages print. Python does not raise an error here — each `if` is a separate, independent check, so a score of 95 satisfies all three conditions and every matching branch runs.

This is a logic bug, not a syntax bug: nothing crashes, so it is easy to miss until the output is checked line by line.

**Correction steps**

1. Decide whether the branches are mutually exclusive (only one should ever apply).
2. If so, connect them with `elif` so Python stops checking after the first true condition.
3. Reserve separate `if` statements for conditions that really are independent of one another.

```python
score = 95

if score >= 90:
    print("Excellent")
elif score >= 70:
    print("Passing")
elif score >= 0:
    print("Needs support")
```

## Common syntax slips to recognize

| Slip | What happens | Fix |
|---|---|---|
| `else if score >= 70:` | `SyntaxError` — Python's keyword is one word | Use `elif`, not two words |
| `if score >= 90` (no colon) | `SyntaxError` — every `if`/`elif`/`else` line needs a trailing `:` | Add the colon |
| `if score = 90:` | `SyntaxError` — `=` assigns, `==` compares | Use `==` to test equality |

These three all stop the program immediately, so the traceback is the first clue. Compare the reported line against this table before rereading the whole cell.

### Practice 5: Combine conditions with `or` and negate with `not`

**Reference note:** `and` requires every condition to be true, `or` requires only one, and `not` flips a Boolean value. Combining them lets one branch cover several situations.

**Before running:** Predict the exact output and identify the line responsible for each result.

### Practice 3: Use an elif chain for categories

**Reference note:** Python checks the branches from top to bottom and runs the first true branch. Ordering the thresholds from highest to lowest avoids overlap mistakes.

**Before running:** Predict the exact output and identify the line responsible for each result.

**Lab change:** Test the values immediately below, at, and above each threshold.


```python
score = 84
if score >= 90:
    letter = "A"
elif score >= 80:
    letter = "B"
elif score >= 70:
    letter = "C"
else:
    letter = "Needs review"

print("Category:", letter)
```
