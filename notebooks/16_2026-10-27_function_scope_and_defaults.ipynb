## Conceptual analogy

A local variable is like a note written inside one meeting room. People inside can use it, but someone in the hallway cannot automatically read it. Parameters carry information into the room; a return value carries a result back out.

> This analogy provides an initial mental model. The definitions and Python examples below provide the precise technical meaning.


## Key terms

| Term | Plain-language meaning |
|---|---|
| **scope** | The region where a name is available. |
| **local variable** | A name created inside a function and normally available only there. |
| **default parameter** | A parameter value used when the caller omits that argument. |
| **keyword argument** | An argument supplied with its parameter name. |


## Explanation


Variables created inside a function are local. Avoid relying on global variables when parameters can provide the needed data. Defaults make a common choice convenient, but required parameters must appear before default parameters. Test both the default path and an explicitly supplied alternative.

## Predict before running

The next cell tries to print a local variable after the function ends. What error will appear?

1. Predict the output or error.
2. Record the reason for the prediction.
3. Run the next cell only after recording both parts.


```python
def calculate_tax(price):
    tax = price * 0.08
    return tax

print(tax)
```

### Result and diagnosis

Python raises a `NameError` because `tax` is local to `calculate_tax`. The function returned a value, but the caller did not store it.

**Error-reading procedure:** Read the final line of the message, locate the referenced line, and compare the value or structure Python received with the program's intended behavior.


### Correction steps

1. Call the function with an argument.
2. Store the returned value in a name available to the caller.
3. Print or reuse that caller-side name.


```python
def calculate_tax(price, rate=0.08):
    tax = price * rate
    return tax

default_rate_tax = calculate_tax(100)
custom_rate_tax = calculate_tax(100, rate=0.05)
print("Tax with example default rate:", default_rate_tax)
print("Tax with custom rate:", custom_rate_tax)
```

### How the corrected example works

1. `tax` exists only during each function call.
2. The first call uses the default rate.
3. The second call uses a keyword argument to make the alternative rate clear.

A useful verification step is to change one value, predict the new result, and run the cell again.

## Additional in-class code lab

### Practice 1: Use a default parameter value

**Reference note:** A default supplies an argument when the caller omits it. Passing an explicit value overrides the default for that call only.

**Before running:** Predict the exact output and identify the line responsible for each result.

**Lab change:** Replace the default and verify which calls change.


```python
def describe_origin(country="Norway"):
    print("Country:", country)

describe_origin()
describe_origin("India")
describe_origin("United States")
```

### Practice 2: Call parameters by keyword

**Reference note:** Keyword arguments connect each value to a parameter name, so the call remains clear even when the order changes.

**Before running:** Predict the exact output and identify the line responsible for each result.

**Lab change:** Call the function once with positional arguments and once with all keyword arguments.


```python
def student_label(name, age, program="CSCI"):
    return name + " | " + str(age) + " | " + program

print(student_label("Jordan", 20))
print(student_label(program="Data Science", age=22, name="Mina"))
```

### Practice 3: Distinguish local and global names

**Reference note:** A name created inside a function is local to that call. The separate name outside the function remains available after the call.

**Before running:** Predict the exact output and identify the line responsible for each result.

**Lab change:** Create another local name and predict whether code after the function can use it.


```python
course_name = "CSCI 171"

def show_course():
    local_message = "Function practice"
    print(course_name)
    print(local_message)

show_course()
print("Outside:", course_name)
```

### Practice 4: Test a function at boundaries

**Reference note:** A function is not finished after one successful call. Tests below, at, and above the threshold expose comparison mistakes.

**Before running:** Predict the exact output and identify the line responsible for each result.

**Lab change:** Add tests for 0 and 100 and explain what they verify.


```python
def is_passing(score, minimum=70):
    return score >= minimum

print("69:", is_passing(69))
print("70:", is_passing(70))
print("71:", is_passing(71))
print("Custom minimum:", is_passing(84, minimum=85))
```
